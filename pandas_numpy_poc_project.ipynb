## Outline

It heavily utilizes both **Pandas** (for time-series data wrangling) and **NumPy** (for fast, vectorized mathematical operations), and it ties directly into Imperial’s core module: **Investments and Portfolio Management**.

### Project Title: Quantitative Portfolio Optimizer & Monte Carlo Simulator
**The Goal:** Build a script that takes a basket of historical stock data, calculates their risk and return metrics, and uses a Monte Carlo simulation (generating thousands of random portfolios) to find the mathematically optimal allocation of assets (the "Efficient Frontier"). 


### Phase 1: Time-Series Data Wrangling (Heavy Pandas)
In traditional finance, you use monthly or daily returns, not raw prices. Your first job is to clean the data so it's ready for mathematical modeling. 
*   **The Task:** Download 5 years of daily closing prices for 5 different stocks (e.g., AAPL, MSFT, JPM, TSLA, and a Treasury Bond ETF) using a free library like `yfinance`. 
*   **Key Pandas Functions to Master:**
    *   `pd.concat()` or `.join()`: To merge the 5 different price series into one single DataFrame, aligned by the Date index.
    *   `.fillna()` and `.dropna()`: Financial data always has missing days (holidays, trading halts). You must clean these.
    *   `.pct_change()`: **Crucial.** Converts raw daily prices into daily percentage returns. 
    *   `.resample('M').last()`: To convert your daily data into monthly data, which is standard for long-term portfolio analysis.
    *   `.rolling(window=252)`: To calculate a rolling 1-year (252 trading days) moving average or rolling volatility.

### Phase 2: Statistical Risk Metrics (Pandas + NumPy)
Imperial’s *Financial Econometrics* module will focus heavily on variance, covariance, and correlation. You will calculate these using a mix of both libraries.
*   **The Task:** Calculate the historical annualized returns, annualized volatility (risk), and how the stocks move in relation to one another.
*   **Key Functions to Master:**
    *   `np.log(1 + returns)`: Using NumPy to calculate logarithmic returns, which are preferred in quantitative finance over simple returns.
    *   `df.cov()` and `df.corr()`: Pandas functions to generate a Covariance Matrix and a Correlation Matrix. (Understanding the Covariance matrix is fundamental to FinTech).
    *   `np.sqrt()`: Used in conjunction with variance to calculate standard deviation (volatility).

### Phase 3: The Monte Carlo Simulation (Heavy NumPy)
This is where you leave basic Business Analytics behind and enter quantitative FinTech. Instead of guessing the best weights (e.g., 20% in each of the 5 stocks), you will simulate 10,000 different random combinations to find the best one. 
*   **The Task:** Generate 10,000 random portfolios, calculate the expected return and risk for each, and identify the one with the highest Sharpe Ratio (best risk-adjusted return).
*   **Key NumPy Functions to Master:**
    *   `np.random.random()`: To generate random arrays of weights for your 5 stocks.
    *   `weights / np.sum(weights)`: Array broadcasting. Ensures your random weights always equal 100% (or 1.0).
    *   `np.dot()`: **The most important NumPy function for finance.** You will use matrix multiplication (the dot product) to multiply your array of weights by your Covariance Matrix to calculate the total portfolio risk.
    *   `.T` (Transpose): You will frequently need to transpose NumPy arrays (`weights.T`) to make the linear algebra math work.

### Phase 4: Algorithmic Rebalancing (Advanced Pandas)
If you buy a portfolio today, stock prices change, and your weights drift. A true FinTech application tests what happens if you algorithmically "rebalance" the portfolio back to its target weights every month.
*   **The Task:** Simulate holding your optimal portfolio over the 5 years, automatically rebalancing it on the 1st of every month.
*   **Key Pandas Functions to Master:**
    *   `df.shift(1)`: To prevent "look-ahead bias" (a cardinal sin in algorithmic trading). You must shift your data down one row so your algorithm only makes decisions based on *yesterday's* data, not today's.
    *   `df.groupby(pd.Grouper(freq='M'))`: To isolate the final trading day of every month to trigger your rebalancing code.
    *   `.cumprod()`: Cumulative product, used to calculate how your initial $10,000 investment grows over the 5-year timeline.
